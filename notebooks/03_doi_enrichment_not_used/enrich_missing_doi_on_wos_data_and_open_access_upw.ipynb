# import data

```python
wos_data_and_open_access_upw = pd.read_pickle('D:\RESEARCH IN AFRICA\Manuscripts/Open access paper/data_analysis/wos_data_and_open_access_upw.pkl')
wos_data_and_open_access_upw = wos_data_and_open_access_upw.applymap(lambda s: s.lower() if type(s) == str else s)
wos_data_and_open_access_upw
```

```python
add_doi_data_from_crossref = pd.read_pickle(f'../../data/add_doi_data_from_crossref.pkl')[['wos_ID','crossref_ID','doi']].dropna(subset=['doi'], axis=0)
add_doi_data_from_crossref
```

```python
add_doi_data_from_scopus = pd.read_pickle(f'../../data/add_doi_data_from_scopus.pkl')[['wos_ID','scopus_ID','doi']].dropna(subset=['doi'], axis=0)
add_doi_data_from_scopus
```

```python
add_doi_data_from_lens = pd.read_pickle(f'../../data/add_doi_data_from_lens.pkl')[['wos_ID','lens_ID','doi']].dropna(subset=['doi'], axis=0)
add_doi_data_from_lens
```

```python
add_doi_data_concat = pd.concat([pd.concat([add_doi_data_from_scopus.set_index('wos_ID'),add_doi_data_from_lens.set_index('wos_ID')]),
                       add_doi_data_from_crossref.set_index('wos_ID')])
add_doi_data_concat = add_doi_data_concat.rename(columns={'doi':'add_doi'}).drop_duplicates('add_doi')
add_doi_data_concat
```

```python
wos_data_and_open_access_upw_and_add_doi = wos_data_and_open_access_upw.merge(add_doi_data_concat[['add_doi']].reset_index(),on='wos_ID',how='left').drop_duplicates('wos_ID')
wos_data_and_open_access_upw_and_add_doi
```

```python
wos_data_and_open_access_upw_and_add_doi.count()
```

```python
wos_data_and_open_access_upw_and_add_doi['doi'] = wos_data_and_open_access_upw_and_add_doi['doi'].fillna(wos_data_and_open_access_upw_and_add_doi['add_doi'])
wos_data_and_open_access_upw_and_add_doi#.count()
```
