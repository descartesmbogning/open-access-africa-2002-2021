```python
data_prep_all_funding = data_prep_all[['UT (Unique WOS ID)','Funding Orgs']].rename(columns={'UT (Unique WOS ID)':'wos_ID'})
data_prep_all_funding['wos_ID'] = data_prep_all_funding['wos_ID'].str.lower()

# Applying the condition
data_prep_all_funding["Funding Orgs"] = data_prep_all_funding["Funding Orgs"].fillna(0)

data_prep_all_funding.loc[data_prep_all_funding["Funding Orgs"] == 0, "funding_yes"] = 'not_fund'
data_prep_all_funding.loc[data_prep_all_funding["Funding Orgs"] != 0, "funding_yes"] = 'fund'

data_prep_all_funding
```

```python
data_prep_all_Categories = data_prep_all[['UT (Unique WOS ID)','WoS Categories']].rename(columns={'UT (Unique WOS ID)':'wos_ID'})
data_prep_all_Categories['wos_ID'] = data_prep_all_Categories['wos_ID'].str.lower()
```

```python
data_concat_Categories = data_concat.merge(data_prep_all_Categories, on = 'wos_ID', how='left')
data_concat_Categories
```

```python
data_concat_Categories_table = data_concat_Categories.groupby(['WoS Categories']).count()

#save data
data_concat_Categories_table.to_csv('D:\RESEARCH IN AFRICA\Manuscripts/research strend in Africa wos/data/categories_table_count_countries.csv')

data_concat_Categories_table
```

```python
# categories year
```

```python
data_concat_categories2 = data_concat_Categories[['year','year_group','WoS Categories']].set_index(['year_group','year'])
data_concat_categories3 = data_concat_categories2['WoS Categories'].str.split('; ', expand=True)
data_concat_categories3
```

```python
# get dummies to have contries on columns
from functools import reduce
dfs = [pd.get_dummies(data_concat_categories3[col]) for col in data_concat_categories3]
d = reduce(lambda x,y : x.add(y, fill_value=0), dfs)

# data cleaning
cols = [c for c in d.columns if c[-1:] != '.']
df=d[cols]
cols = [c for c in df.columns.to_list() if len(c) != 1]
df=df[cols]

dfs= df.replace(0.0, np.nan)
#save data
dfs.to_csv('D:\RESEARCH IN AFRICA\Manuscripts/research strend in Africa wos/data/categories_table_count.csv')

dfs
```

```python
data_concat_categories_year_table = dfs.groupby(['year']).count()
#save data
data_concat_categories_year_table.to_csv('D:\RESEARCH IN AFRICA\Manuscripts/research strend in Africa wos/data/categories_table_count_years.csv')

data_concat_categories_year_table
```

```python
data_concat_categories_year_g_table = dfs.groupby(['year_group']).count()
#save data
data_concat_categories_year_g_table.to_csv('D:\RESEARCH IN AFRICA\Manuscripts/research strend in Africa wos/data/categories_table_count_years_group.csv')

data_concat_categories_year_g_table
```
