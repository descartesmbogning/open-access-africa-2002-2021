# 1. What percentage of OA in general and according to year and countries

## group by 'is_oa' 

### all

```python
data_countries_grp_is_oa = data_countries.groupby(['is_oa']).count().T.iloc[4:]
data_countries_grp_is_oa
```

```python
data_countries_count = data_countries.count().to_frame().iloc[5:]
data_countries_count
```

```python
for col in data_countries_grp_is_oa.columns:
    for row in data_countries_grp_is_oa.index.to_list():
        data_countries_grp_is_oa[col] = data_countries_grp_is_oa[col]
        data_countries_grp_is_oa[col+'_%'] = data_countries_grp_is_oa[col]/data_countries_count[0]*100
        data_countries_grp_is_oa[col+'_%_z_score'] = (data_countries_grp_is_oa[col+'_%'] - data_countries_grp_is_oa[col+'_%'].mean()) /data_countries_grp_is_oa[col+'_%'].std()

#save data
data_countries_grp_is_oa.to_csv('D:\RESEARCH IN AFRICA\Manuscripts/Open access paper/data_analysis/tables/data_countries_grp_is_oa_z_score_african_languages.csv')
data_countries_grp_is_oa
```
