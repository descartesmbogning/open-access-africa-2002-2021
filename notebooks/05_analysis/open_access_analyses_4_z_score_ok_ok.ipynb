## group by 'is_oa' 

```python
data_oa_funding = data_oa[['wos_ID','is_oa','funding_yes']].set_index(['wos_ID','funding_yes'])
data_oa_funding
```

```python
# get dummies to have contries on columns
from functools import reduce
dfs = [pd.get_dummies(data_oa_funding[col]) for col in data_oa_funding]
d = reduce(lambda x,y : x.add(y, fill_value=0), dfs)

# data cleaning
cols = [c for c in d.columns if c[-1:] != '.']
df=d[cols]
cols = [c for c in df.columns.to_list() if len(c) != 1]
df=df[cols]

dfs= df.replace(0.0, np.nan).reset_index()
dfs
```

```python
data_oa_grp_funding = dfs.groupby(['funding_yes']).count()
data_oa_grp_funding
```

```python
for col in data_oa_grp_funding.columns:
    for row in data_oa_grp_funding.index.to_list():
        data_oa_grp_funding[col] = data_oa_grp_funding[col]
        data_oa_grp_funding[col+'_%'] = data_oa_grp_funding[col]/data_oa_grp_funding['wos_ID']*100
        data_oa_grp_funding[col+'_%_z_score'] = (data_oa_grp_funding[col+'_%'] - data_oa_grp_funding[col+'_%'].mean()) /data_oa_grp_funding[col+'_%'].std()

#save data
data_oa_grp_funding.to_csv('D:\RESEARCH IN AFRICA\Manuscripts/Open access paper/data_analysis/tables/data_oa_grp_funding_z_score.csv')
data_oa_grp_funding
```
