```python
# funding countries
```

```python
data_prep_all_oa = data_prep_all_keep.rename(columns={'UT (Unique WOS ID)':'wos_ID'})
data_prep_all_oa['wos_ID'] = data_prep_all_oa['wos_ID'].str.lower()
data_prep_all_oa['DOI'] = data_prep_all_oa['DOI'].str.lower()

# Applying the condition
data_prep_all_oa["Funding Orgs"] = data_prep_all_oa["Funding Orgs"].fillna(0)

data_prep_all_oa.loc[data_prep_all_oa["Funding Orgs"] == 0, "funding_yes"] = 'not_fund'
data_prep_all_oa.loc[data_prep_all_oa["Funding Orgs"] != 0, "funding_yes"] = 'fund'

data_prep_all_oa
```

```python
# merge
data_concat_oa = wos_data_countries[['wos_ID']].merge(data_prep_all_oa, on = 'wos_ID', how='left')
data_concat_oa
```

```python
#save data
data_concat_oa.to_pickle('D:\RESEARCH IN AFRICA\Manuscripts/Open access paper/data_analysis/wos_infos_for_open_access_africa_verif.pkl')
```
